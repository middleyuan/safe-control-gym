## ID param

```python
noise_type = 'param'
id = np.load(f'{script_dir}/../data/gpmpc_acados_TP_param_{noise_type}_results.npy', allow_pickle=True).item()
nominal = np.load(f'{script_dir}/../data/gpmpc_acados_TP_{noise_type}_results.npy', allow_pickle=True).item()
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(nominal['noise_factor'], nominal['rmse_degradation_mean'], 
        label='Nominal', color='tab:grey',)
ax.fill_between(nominal['noise_factor'], 
                nominal['rmse_degradation_mean'] - nominal['rmse_degradation_std'], 
                nominal['rmse_degradation_mean'] + nominal['rmse_degradation_std'], 
                alpha=0.2, color='tab:grey')

ax.plot(id['noise_factor'], id['rmse_degradation_mean'], 
        label='ID', color='tab:blue',)
ax.fill_between(id['noise_factor'], 
                id['rmse_degradation_mean'] - id['rmse_degradation_std'], 
                id['rmse_degradation_mean'] + id['rmse_degradation_std'], 
                alpha=0.2, color='tab:blue')

# make y tick percentage in the plot and stick to data
ax.set_yticks(ax.get_yticks())
ax.set_yticklabels([f'{int(y)}%' for y in ax.get_yticks()])

ax.set_xlabel('Noise amplification factor')
ax.set_ylabel('RMSE degradation')
ax.set_ylim([0, None])

ax.legend()
ax.set_title(f'RMSE degradation of GP-MPC with {noise_type}')
plt.tight_layout()
```
